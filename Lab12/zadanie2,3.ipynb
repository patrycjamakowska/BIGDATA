```
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
import random

data = [(i, random.choice(["PL", "DE", "US", "FR"])) for i in range(1000)]
df = spark.createDataFrame(data, ["id", "country"])

df.write.mode("overwrite") \
    .partitionBy("country") \
    .parquet("/tmp/example_partitioned")

df.write \
  .format("parquet") \
  .mode("overwrite") \
  .bucketBy(4, "country") \
  .sortBy("country") \
  .saveAsTable("bucketed_table")

display(dbutils.fs.ls("dbfs:/tmp/example_partitioned"))
display(spark.sql("SHOW TABLES").filter("tableName = 'bucketed_table'"))
```
