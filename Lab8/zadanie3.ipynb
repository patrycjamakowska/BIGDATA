```
#do sprwadzenia czy mamy potrzebne kolumny
def columns(data, col=None):
    if col is None:
        required_columns = {"id","name","value"}
    missing = required_columns - set(data.columns)
    if missing:
        raise ValueError("Bad columns")
```

```
#brakujące wartości
def nulls(df):
    if df.isnull().values.any():
        raise ValueError("Null values")
```

```
#sprawdzenie typow kolumn
def column_type(df, expected_types):
    for col, expected_type in expected_types.items():
        if expected_type == "int" and not is_integer_dtype(df[col]):
            raise TypeError("Bad type")
        if expected_type == "float" and not is_float_dtype(df[col]):
            raise TypeError("Bad type")
        if expected_type == "datetime" and not is_datetime64_any_dtype(df[col]):
            raise TypeError("Bad type")
```

```
#sprawdzenie dupliakatow
def unique_column(df, col):
    if df[col].duplicated().any():
        raise ValueError("This column have duplicates")
```
