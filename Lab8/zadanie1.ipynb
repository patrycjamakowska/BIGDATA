```
df.printSchema()
```

```
from pyspark.sql.functions import explode, col
```

```
flat= df.select(explode("features").alias("feature")).select(
    col("feature.geometry"),
    col("feature.geometry.type").alias("geometryType"),
    col("feature.geometry.coordinates").alias("coordinates"),
    col("feature.properties.featureGroup").alias("featureGroup"),
    col("feature.properties.featureId").alias("featureId"),
    col("feature.properties.jobReference").alias("jobReference"),
    col("feature.properties.baseFormComponent.componentId").alias("baseFormComponentId"),
    col("feature.properties.baseFormComponent.form").alias("baseFormForm"),
    col("feature.properties.baseFormComponent.versionCreationMetadata.editDate").alias("baseFormEditDate"),
    col("feature.properties.lifecycleStatusComponent.lifecycleStatus").alias("lifecycleStatus"),
    col("feature.properties.lifecycleStatusComponent.versionCreationMetadata.editDate").alias("lifecycleEditDate")
)

display(flat)
```
